```python
class Employee:

    id:int
    name:str
    department:str
    salary:int

    def set_attribute(self,id,name,department,salary):

        self.id = id
        self.name = name
        self.department = department
        self.salary = salary

    def display_employee(self):
        print(self.id,self.name,self.department)

emp_instance1 = Employee()

emp_instance2 = Employee()

emp_instance1.set_attribute(100,"rahul","hr",350000)

emp_instance1.display_employee()


# self => current instance point cheyyaaan
```

## Constructor

--- 
- **is a special methods that initilaize attribute of a class**
             
- **name of constructor will be** **` __init__`**

- **constructor automatically invoked during object creation**



```python
class Student:

    roll_num = int

    name:str

    course:str

    def __init__(self,roll_num,name,course):

        self.roll_num = roll_num

        self.name = name

        self.course = course


    def display_student(self):

        print("Roll Num =",self.roll_num,", Name =",self.name,", Course =",self.course)


student_instance = Student(1,"Abhishek","Django")

student_instance.display_student()
        
```

## OOP Features

- **Inheritance** => **Child class accessing properties and methods of its parent class**


```python
class Parent:

    def home(self):

        print("Home....")

class Child(Parent):

    def social_media(self):

        print("insta","youtube")

child_instance = Child()

child_instance.social_media()

child_instance.home()
```

- **Multiple Inheritance** => **A child class inheritant from multiple inheritance**

```python
class Person:

    def eat(self):
        print("eating.....")

    def walk(self):
        print("walking...")

class CricketPlayer:

    def bat(self):
        print("bating....")

class Developer(Person,CricketPlayer):

    def code(self):
        print("coding...")

    def debug(self):
        print("debugging...")

developer_instance = Developer()

developer_instance.code()

developer_instance.debug()

developer_instance.eat()

developer_instance.bat()
```
