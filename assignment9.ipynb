# Expense Dataset


```python
expenses = [
    {"id": 1, "title": "Groceries", "amount": 2500, "category": "Food", "payment_method": "UPI", "owner": "Rahul"},
    {"id": 2, "title": "Internet Bill", "amount": 999, "category": "Bills", "payment_method": "Credit Card", "owner": "Anjali"},
    {"id": 3, "title": "Electricity Bill", "amount": 3500, "category": "Bills", "payment_method": "UPI", "owner": "Rahul"},
    {"id": 4, "title": "Movie Tickets", "amount": 800, "category": "Entertainment", "payment_method": "Debit Card", "owner": "Priya"},
    {"id": 5, "title": "Dining Out", "amount": 1800, "category": "Food", "payment_method": "Credit Card", "owner": "Rahul"},
    {"id": 6, "title": "Gym Membership", "amount": 1500, "category": "Health", "payment_method": "UPI", "owner": "Anjali"},
    {"id": 7, "title": "Fuel", "amount": 2000, "category": "Travel", "payment_method": "Cash", "owner": "Priya"},
    {"id": 8, "title": "Books", "amount": 650, "category": "Education", "payment_method": "UPI", "owner": "Rahul"},
]
```

 1. Extract a list of all expense titles.

```python
all_expense_title = [e.get("title") for e in expenses]

print(all_expense_title)
```

2. Extract a list of all expense amounts.

```python
all_expense_amounts = [e.get("amount") for e in expenses]
print(all_expense_amounts)
```

17. Find the maximum expense amount using max() and list comprehension.

```python
all_expense_amounts = [e.get("amount") for e in expenses]

max_expense_amount = max(all_expense_amounts)


print(max_expense_amount)
```

```python
def get_expenses(e):
    return e.get("amount")

print(max(expenses,key=get_expenses).get("amount"))
```

18. Find the minimum expense amount using min() and list comprehension.

```python
all_expense_amounts = [e.get("amount") for e in expenses]

min_expense_amount = min(all_expense_amounts)


print(min_expense_amount)
```

```python
def get_expenses(e):
    return e.get("amount")

print(min(expenses,key=get_expenses).get("amount"))
```

19. Get the title of the expense that has the highest amount using list comprehension and max().

```python
all_expense_amounts = [e.get("amount") for e in expenses]

max_expense_amount = max(all_expense_amounts)


max_expense_title = [e.get("title") for e in expenses if e.get("amount") == max_expense_amount]
print(max_expense_title)
```

```python
def get_expenses(e):
    return e.get("amount")

print(max(expenses,key=get_expenses).get("title"))
```

20. Calculate the total sum of all expenses spent by 'Rahul' using list comprehension and sum().
