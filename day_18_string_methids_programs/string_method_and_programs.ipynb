## string and string methods

### string indexing

find particular element in sequence or string

```

text="hello"
      01234

print(text[0]) =>h
print(text[1]) =>e

print(text[-1]) =>o

````python
text="python"

#+ve indexing

print(text[0])
print(text[5])

#-ve indexing

print(text[-1])
print(text[-5])
````


`index(value)` => return index of first occurance of value

eg:

    ```
    word="hello"

    pos=word.index("l") =>2

    ```

````python
text="luminar"

pos=text.index("m")

print(pos)
````

`find(value)` => return index position of first occurance of value return -1 if value not exist in str

eg:

    ```
       text="hello"

       pos=text.find("l") =>3

       print(pos)

    ```

````python
text="python"

pos=text.find("u") #=> -1 bcz u value not exist in text string
pos=text.find("h") #=> 3

print(pos)
````

`rfind(value)` =>return index position of last occurance of value return -1 if value not exist 

```
  "hello".rfind("l") =>3

````python
text="hello"

pos=text.rfind("l")

print(pos)
````

`replace(old,new,count)` => **replace old value with new value**

````python
text="hello"

new_text=text.replace('l',"L")

new_text
````

````python
text="hello"

new_text=text.replace('l',"L",count=1)

new_text
````

`strip()` => **remove value from both end**

````python
text="###hello###"

new_text=text.strip("#")

new_text
````

````python
text="\thello\t"

new_text=text.strip("\t")

print(new_text)
````

`lstrip(value)` => **remove value from left** 

`rstrip(value)` => **remove value from right**

````python
text=",hello"

new_text=text.lstrip(",")
new_text
````

````python
text=",hello,"

new_text=text.rstrip(",")
new_text
````

`starswith(prefix)` => **return true if the string start with the specified prefix , false otherwise**

````python
text="python"

text.startswith("py")
````

`endswith(suffix)` => **return true if the string ends with the specified suffix , false otherwise**

````python
text.endswith("o")
````

`count(value)` => **return frequency of value in string object**

````python
text="hello"

freq=text.count("l")

freq
````

`split(value)` => **break a string into smaller parts and returns them as a list.**
                  **By default, it splits wherever there is whitespace.**

````python
text="hello,hai,how,are,you"

words=text.split(",")

words
````

````python
data="10,rahul,kakkanad,ceo"

id,name,place,dept=data.split(",")

print(id)
print(name)
````

````python
text = "Hello World"
print(text.split())
````
