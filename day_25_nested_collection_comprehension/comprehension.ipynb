## Comprehension

**easy way creating list|set|dic from sequence**


---
syntax:

    ist=[return_value,iteration,condition]


---

eg:
   lst=[10,20,30,40,50]

   add=[i+5 for i in lst]
   
   print(add)

   OUTPUT:
       [15,25,35,45,55]

```python
lst=[10,20,30,40,50]

squares = [i**2 for i in lst]

print(squares)
```

```python
lst=[10,20,30,40,50]

add=[i+5 for i in lst]

print(add)
```

```python
lst=[10,20,30,40,50]

cube = [i**3 for i in lst]

print(cube)
```

```python
double = [i*2 for i in lst]

print(double)
```

```python
#greater than 13

lst = [10,11,12,13,14]

gtr_13 = [num for num in lst if num>13]

print(gtr_13)
```

```python
# even number
lst = [10,11,12,13,14]

evens = [num for num in lst if num%2==0]

print(evens)
```

```python
words=["flight","flower","fly","fluid"]

upper_words = [ch.upper() for ch in words]

print(upper_words)
```

```python
# return words whose length > 3

grt_3 = [w for w in words if len(w)>3]

print(grt_3)
```

```python
lst = [10,11,13,12,11,12,10,11,10,11,13,14]

count_lst = {num:lst.count(num) for num in lst}

print(count_lst)
```

```python
squares = {num:num**2 for num in lst}

print(squares)
```

```python
# common prefix

words = ["flight","flower","fly","fluid"]


words.sort()

first = words[0]
last = words[-1]

small_word = first if len(first) < len(last) else last

prefix = ""

for i in range(len(small_word)):

    if first[i]==last[i]:

        prefix=prefix+first[i]

    else:

        break

print(prefix)
```
