## list methods

- `index(value)` => **return index of value**
- `remove(value)` => **remove first occurance of value from list**

```python
arr = [10,20,30,20,40]

print(arr.index(20))   # index(value)

arr.remove(20)   # remove(value)

print(arr)
```

- `count(value)` => **return frequency of value from list**

```python
arr = [10,20,20,10,11,12,11,13,14]

print(arr.count(10))

for num in arr:
    print(num,arr.count(num))
```

- `sort()`  => **sort list**

```python
# Asc order
arr = [3,1,5,2,7,5,8]

arr.sort()

print(arr)
```

```python
# disc order
arr = [3,1,5,2,7,5,8]

arr.sort(reverse=True)

print(arr)
```

- `reverse()` => **reverse the list in place**

```python
chars = ["t","a","n"]

chars.reverse()

print(chars)
```
