# Tuple 

**ordered,allow duplicates but immutable**

`tuple can define with`()



```python
course = ("django","mern","testing")

course[0] = "datascience"  # error bcz tuple is immutable
```

## tuple methods

- `index(value)` => **return index position of value**

- `count(value)` => **return frequency of value**



```python
course = ("django","mern","testing","django")

print(course.index("mern")) #1

print(course.count("django")) #2
print(course.count("mern")) #1
```
