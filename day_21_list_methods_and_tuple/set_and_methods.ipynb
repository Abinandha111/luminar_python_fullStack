## Set 

**unordered,unique(not allowed duplicates) and mutable**

- define set with `st=set() or st={1,2,3}`

```python
st={10,20,30}

print(st[10]) #error

st[1]=100 #error
```

```python
st={10,20,30,40,50,40}

print(st)
```

## set methods

- `add(value)` => **add values**


```python
st={10,20,30,40,50,40}

st.add(60)
print(st)
```

- `union(sequence)` => **combine two or more sets**

```python
s1={10,20,30,40,50}

s2={10,20,100,200,300,400}

union_set = s1.union(s2)

print(union_set)
```

- `intersecion(seq)` => **common**

```python
s1={10,20,30,40,50}

s2={10,20,100,200,300,400}

intersection_set = s1.intersection(s2)

print(intersection_set)
```

- `difference(seq)`

```python
s1={10,20,30,40,50}

s2={10,20,100,200,300,400}

diff_set = s1.difference(s2)

print(diff_set)
```

```python
s1={10,20,30,40,50}

s2={10,20,100,200,300,400}

diff_set = s2.difference(s1)

print(diff_set)
```

- ## issuperset()



```python
s1={10,20,30,40,50}
s2={10,20,30}

print(s1.issuperset(s2))
```

- ## issubset()

```python
s1={10,20,30,40,50}
s2={10,20,30}

print(s2.issubset(s1))
```
