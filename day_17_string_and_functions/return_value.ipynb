## Function for adding two numbers







```python
def add_numbers(n1,n2):
    result= n1 + n2

    return result

value=add_numbers(100,200)
print(value)
```

## return cube of a number

```python
def cube_number(n):
    result = n**3

    return result

value=cube_number(2)

print(value)
```

## is a num odd

```python
def is_odd(n):

    if n%2!=0:

        return True

    else:

        return False

value=is_odd(3)

print(value)
```
