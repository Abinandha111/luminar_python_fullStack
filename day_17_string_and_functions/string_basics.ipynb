### python provide string methods to handle string object

**case conevrsion method**

`lower()` => convert to lower case

`upper()` => convert to upper case

`isdigit()` => return true if string object contain only digit return false otherwise

`isalpha()` => return true if string object contain only alphabet return false otherwise

`isalnum()` => return true if string object contain only alphabet and digit




```python
name="rahul"

upper_case=name.upper()

print(upper_case)
```

```python
employee_name="RAHUL"

lower_case=employee_name.lower()

print(lower_case)
```

```python
vehicle_number="10110"

is_number=vehicle_number.isdigit()

print(is_number)
```

```python
challan_number="hhdgsgshgyeed"

is_alphabet=challan_number.isalpha()

print(is_alphabet)
```

```python
pan_card_number="dashashdh373673"

print(pan_card_number.isalnum())
```

### write a program to check pancard number is valid

```python
pan_card_number=input("Enter pan card number:")

if pan_card_number.isalnum() and len(pan_card_number)==10:

    print("valid")

else:

    print("invalid")
```
