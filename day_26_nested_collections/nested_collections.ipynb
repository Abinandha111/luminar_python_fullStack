```python
#list of dictionary

movies = [

    {"id":1,"title":"kgf","year":2018,"run_time":156,"rating":9},
    {"id":2,"tile":"kgf2","year":2023,"run_time":167,"rating":9.5},
    {"id":3,"title":"arm","year":2024,"run_time":150,"rating":8.5},
    {"id":4,"title":"premalu","year":2017,"run_time":166,"rating":10},
    {"id":5,"title":"thudarum","year":2025,"run_time":170,"rating":8},
]

#kgf2 runtime

print("kgf runtime ",movies[1].get("run_time"))

#thudarum year 

print("thudarum year ",movies[-1].get("year"))

#premalu rating

print("premalu rating ",movies[-2].get("rating"))

```

## filter

```python
#movies after 2018

year = [di.get("title") for di in movies if di.get("year")>2018]

print(year)
```

```python
#movies before 2018

movies_before_2018 = [di.get("title") for di in movies if di.get("year")<2018]

print(movies_before_2018)
```

```python
# movies with rating > 9

rating_filter = [di.get("title") for di in movies if di.get("rating")>9]

print(rating_filter)
```

```python
# movies with run_time > 158

run_time_filter = [di.get("title") for di in movies if di.get("run_time")>158]

print(run_time_filter)
```

```python
all_rating = [di.get("rating") for di in movies ] #MAPPING

max_rating = max(all_rating) #REDUCE

top_movies = [di.get("title") for di in movies if di.get("rating") == max_rating]

print(top_movies)
```

```python
# movies with lowest rating

all_rating = [di.get("rating") for di in movies ] #MAPPING

min_rating = min(all_rating) #REDUCE

lowest_movies = [di.get("title") for di in movies if di.get("rating")==min_rating]

print(lowest_movies)
```

```python
# movies with highest run_time

all_runtime = [di.get("run_time") for di in movies]

max_runtime = max(all_runtime)

top_movie = [di.get("title") for di in movies if di.get("run_time") == max_runtime]

print(top_movie)
```

```python
#movies with lowest run_time
min_runtime = min(all_runtime)

movie = [di.get("title") for di in movies if di.get("run_time") == min_runtime]

print(movie)
```

```python
# lowest rating

def get_rating(di):

    return di.get("rating")

print(max(movies,key=get_rating))
```

```python
#higest rating


print(min(movies,key=get_rating))
```

```python
#highest runtime

def get_run_time(di):

    return di.get("run_time")

print(max(movies,key=get_run_time))
```
