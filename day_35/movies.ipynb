```python
class Movies():

    def __init__(self):

        self.movie_records = [
        {"id": 1, "title": "Leo", "actor": "Vijay", "year": 2023}
    ]

    def post(self,**kwargs):

        form_data = kwargs

        self.movie_records.append(form_data)

        print("added....")

    def get(self):
        for di in self.movie_records:
            print(di)

    def retrieve(self,id=None):

        movie = [di for di in self.movie_records if di.get("id") == id][0]

        print(movie)

    def put(self,id=None,**kwargs):

        form_data = kwargs

        movie = [di for di in self.movie_records if di.get("id") == id][0]

        movie.update(form_data)

    def delete(self,id=None):

        movie = [di for di in self.movie_records if di.get("id") == id][0]

        self.movie_records.remove(movie)

    
```

```python
movie_instance = Movies()

movie_instance.post(id=2,title="Premalu",actor="Naslen",year=2024)
```

```python
movie_instance.post(id=3,title="Manjummel Boys",actor="Soubin Shahir",year=2024)
```
