```python
class Expense():

    def __init__(self):

        self.expenses = [
        {"id":1,"title":"Food","amount":250,"Category":"Food","Date":"1-10-2026"}
    ]

    def post(self,**kwargs):

        form_data = kwargs

        self.expenses.append(form_data)

        print("added....")

    def get(self):
        for di in self.expenses:
            print(di)

    def retrieve(self,id=None):

        expense = [di for di in self.expenses if di.get("id") == id][0]

        print(expense)

    def put(self,id=None,**kwargs):

        form_data = kwargs

        expense = [di for di in self.expenses if di.get("id") == id][0]

        expense.update(form_data)

    def delete(self,id=None):

        expense = [di for di in self.expenses if di.get("id") == id][0]

        self.expenses.remove(expense)

        

        


    
```

```python
expense_instance = Expense()

expense_instance.post(id=2,title="Travel",amount=150,Category="Travel",date="4-11-2025")
expense_instance.post(id=3,title="Movie",amount=150,Category="Entertaiment",date="3-9-2025")
```

```python
expense_instance.get()
```

```python
expense_instance.retrieve(id=2)
```

```python
expense_instance.put(id=1,amount=350)
expense_instance.get()
```

```python
expense_instance.delete(id=3)

expense_instance.get()
```
