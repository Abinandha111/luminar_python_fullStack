```python
class Vehicles():

    def __init__(self):

        self.vehicle_records = [
        {"id": 1, "name": "BMW", "type": "Car", "price": 5000000}
    ]

    def post(self,**kwargs):

        form_data = kwargs

        self.vehicle_records.append(form_data)

        print("added....")

    def get(self):
        for di in self.vehicle_records:
            print(di)

    def retrieve(self,id=None):

        vehicle = [di for di in self.vehicle_records if di.get("id") == id][0]

        print(vehicle)

    def put(self,id=None,**kwargs):

        form_data = kwargs

        Vehicle = [di for di in self.vehicle_records if di.get("id") == id][0]

        Vehicle.update(form_data)

    def delete(self,id=None):

        vehicle = [di for di in self.vehicle_records if di.get("id") == id][0]

        self.vehicle_records.remove(vehicle)

    
```

```python
vehicle_instance = Vehicles()

vehicle_instance.post(id=2, name="Royal Enfield", type="Bike", price=250000)
```

```python
vehicle_instance.post(id=3, name="Toyota", type="Car", price=3000000)
```

```python
vehicle_instance.get()
```

```python
vehicle_instance.retrieve(id=3)
```

```python
vehicle_instance.put(id=2,name="Honda", type="Bike")
```
