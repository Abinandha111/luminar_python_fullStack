### Book crud

- **add/create data => post**
- **list/read data => read**
- **update data => put**
- **list one data details => retrieve**
- **delete data => delete**

```python
class Book:

    def __init__(self):
        self.book_records=[
            {"id":1,"title":"randamoozham","author":"mt","price":550,"pages":550}
        ]

    def get(self):
        for di in self.book_records:
            print(di)

    def post(self,**kwargs):
        form_data = kwargs

        self.book_records.append(form_data)

        print("book added...")

    def retrieve(self,id=None):

        book = [di for di in self.book_records if di.get("id") == id][0]

        print(book)

    def put(self,id=None,**kwargs):
        form_data = kwargs

        book = [di for di in self.book_records if di.get("id") == id][0]

        book.update(form_data)

        print("book updated")


    def delete(self,id=None):

        book = [di for di in self.book_records if di.get("id") == id][0]

        self.book_records.remove(book)

        
        print("deleted.....")
```

```python
book_instance = Book()

book_instance.post(id=2,title="Goat life",author="benyamin",price=670,pages=550)
```

```python
book_instance.retrieve(id=1)
```

```python
book_instance.put(id=1,price=780)

book_instance.retrieve(id=1)
```

```python
book_instance.delete(id=1)
```

```python
book_instance.get()
```
