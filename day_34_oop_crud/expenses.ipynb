```python
expenses = [
    {"id":1,"title":"Food","amount":250,"Category":"Food","Date":"1-10-2026"}
]
```

```python
def add_new_expenses(**kwargs):

    form_data = kwargs

    expenses.append(form_data)

add_new_expenses(id=2,title="Travel",amount=120,Category="Transport",Date="3-11-2025")

```

```python
add_new_expenses(id=3,title="Shopping",amount=2000,Category="Groceries",Date="3-11-2024")
```

```python
def list_of_expenses():

    for di in expenses:
        print(di)

list_of_expenses()
```

```python
def delete_expense(id=None):

    expense_details = [di for di in expenses if di.get("id") == id][0]

    expenses.remove(expense_details)
delete_expense(id=1)
```

```python
list_of_expenses()
```

```python
def specific_detail(id=None):
    expense_details = [di for di in expenses if di.get("id") == id][0]
    print(expense_details)
specific_detail(id=2)
```
