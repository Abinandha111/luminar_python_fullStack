```python
vehicles = [
    {"id":1,"name":"Toyota","model":"Fortuner","color":"white","year":2024}
]
```

```python
def adding_vehicles(**kwargs):

    form_data = kwargs

    vehicles.append(form_data)

adding_vehicles(id=2,name="Hyundai",model="Creta",color="black",year=2023)
```

```python
adding_vehicles(id=3,name="Royal Enfield",model="Classic 350",color="Green",year=2024)
```

```python
def list_of_vehicles():

    for di in vehicles:
        print(di)
list_of_vehicles()
```

```python
def sepecific_details(id=None):

    vehicle_detail = [di for di in vehicles if di.get("id") == id][0]

    print(vehicle_detail)
sepecific_details(id=2)
```

```python
def delete_vehicle(id=None):

    vehicle_detail = [di for di in vehicles if di.get("id") == id][0]

    vehicles.remove(vehicle_detail)

    print(vehicles)
delete_vehicle(id=2)
```

```python
list_of_vehicles()
```

```python
def update_details(id=None,**kwargs):

    form_data = kwargs

    vehicle_detail = [di for di in vehicles if di.get("id") == id][0]

    vehicle_detail.update(form_data)

update_details(id=1,name="BMW",model="x5")
```
