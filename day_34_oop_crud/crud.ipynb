```python
employee = [
    {"id":1,"name":"rahul","department":"hr","salary":25000,"location":"tvm"}
]
```

```python
## Add new value

def add_employe(**kwargs):

    form_data = kwargs

    employee.append(form_data)

    print("record inserted")

add_employe(id=2,name="jhon",department="sales",salary=3000,location="tvm")

```

```python
add_employe(id=3,name="jeorge",department="SM",salary=30000,location="ekm")

print(employee)
```

```python
# List all employess

def list_all_employee():

    for di in employee:
        print(di)

list_all_employee()
```

```python
## get Specific employee details

def retrive_employee(id=None):

    employee_detail = [di for di in employee if di.get("id") == id][0]

    print(employee_detail)

retrive_employee(id=2)
```

```python
## delete employee

def remove_employe(id=None):

    emp_data = [di for di in employee if di.get("id") == id][0]

    employee.remove(emp_data)

remove_employe(id=1)
```

```python
list_all_employee()
```

```python
# update employee 

def update_employee(id=None,**kwargs):

    form_data = kwargs

    emplyee_data = [di for di in employee if di.get("id") == id][0]

    emplyee_data.update(form_data)

    print(emplyee_data)

update_employee(id=2,salary=20000,location="tsr")

    


```
