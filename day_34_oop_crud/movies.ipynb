```python
movies = [
    {"id":1,"title":"Inception","director":"Christoper Nolan","language":"English","rating":8.8}
]
```

```python
def adding_movies(**kwargs):

    form_data = kwargs

    movies.append(form_data)

adding_movies(id=2,title="Drishyam",director="Jeethu Joseph",language="Malayalam",rating=8.3)
```

```python
adding_movies(id=3,title="Premam",director="Alphonse",language="Malayalam",rating=9)
```

```python
def list_of_movies():
    for di in movies:
        print(di)
list_of_movies()
```

```python
def delete_movie(id=None):

    movie_detail = [di for di in movies if di.get("id") == id][0]

    movies.remove(movie_detail)
delete_movie(id=3)
```

```python
list_of_movies()
```

```python
def specific_movie(id=None):

    movie_details = [di for di in movies if di.get("id") == id][0]

    print(movie_details)
specific_movie(id=2)
```

```python
def update_movie_details(id=None,**kwargs):

    form_data = kwargs

    movie_detail = [di for di in movies if di.get("id") == id][0]

    movie_detail.update(form_data)

update_movie_details(id=2,title="Interstellar",director="Christoper",language="English")
```
