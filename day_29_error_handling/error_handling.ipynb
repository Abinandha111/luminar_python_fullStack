**accept number and check whether it is odd or even**


```python
try:
    num = int(input("enter a number:"))
except ValueError:
    print("enter  integer value")
else:
    if num%2==0:
        print("even",num)
    else:
        print("odd",num)
```

**accept 2 values from user and calculate average**

```python
try:
    value1=int(input("Enter a value:"))
    value2=int(input("Enter a value:"))
except ValueError:
    print("enter integer value")
else:
    total=value1+value2
    avg=total/2
    print(avg)
```

**accept obtained mark and total mark,calculate percentage and prints it in else blockA**

```python
try:
    mark=int(input("enter your mark:"))
    total_mark=int(input("Enter total mark:"))
    
except ValueError:
    print("Enter integer value")
else:
    percentage=(mark/total_mark)*100
    print(percentage)
```
