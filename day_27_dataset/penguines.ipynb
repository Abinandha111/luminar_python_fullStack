# Lambda function

**anonymous function with single expression**

```python
add = lambda n1,n2:n1+n2
print(add(10,20))

sub = lambda n1,n2:n1-n2
print(sub(20,10))

cube = lambda n:n**3
print(cube(3))

max_two = lambda n1,n2:n1 if n1>n2 else n2
print(max_two(10,20))

min_two = lambda n1,n2:n1 if n1<n2 else n2
print(min_two(10,1))
```

```python
#lowest bill_length

print(min(penguins,key=lambda di:di.get("bill_length")))
```

```python
# highest body_mass

print(max(penguins,key=lambda di:di.get("body_mass")))
```

```python
# lowest body_mass

print(min(penguins,key=lambda di:di.get("body_mass")))
```

```python
# body mass sum

species_body_mass_sum={}

for di in penguins:

    species = di.get("species")
    body_mass = di.get("body_mass")

    if species in species_body_mass_sum:

        species_body_mass_sum[species] += body_mass

    else:

        species_body_mass_sum[species]=body_mass
print(species_body_mass_sum)
```

```python
species_body_mass_sum={di.get("species"):sum([b.get("body_mass") for b in penguins if b.get("species") == di.get("species")]) for di in penguins}

print(species_body_mass_sum)
```
