**create a function to display digits in a given string**

````python


def display_digits(text):

    for ch in text:

        if ch.isdigit():

            print(ch)

display_digits("i have 2 cars,10 bikes and 4 book")
````

**write a function to display vowel count in a string**

````python


def vowel_count(text):

    vowels="aeiouAEIOU"

    count=0

    for ch in text:

        if ch in vowels:
            count=count+1

    print(count)
vowel_count("Hello , Abinandha")

vowel_count("Hello , Abhishek")
````

**Consonants - alphabets that are not vowels**

**write a function that display the count of consonants**


````python
def consonants_counts(text):

    vowels="aeiouAEIOU"

    count=0

    for ch in text:

        if ch not in vowels and ch.isalpha():

            count+=1

    print(count)

consonants_counts("hello1")
consonants_counts("programming ")
````

**Anagram => A word made by rearranging the letters of another word**

**eg:  silent=listen**

**abc=bca**

````python
str1="abc"
str2="bca"

sorted_str1=sorted(str1)

sorted_str2=sorted(str2)

print(sorted_str1==sorted_str2)

    
````

````python
def anagram(text1,text2):

    sorted_str1=sorted(text1)
    sorted_str2=sorted(text2)

    if sorted_str1==sorted_str2:
        print("Anagram")

    else:

        print("not Anagram")

anagram("silent","listen")
````

**pangram => A sentence that contains every letter of the alphabet at least once.**

eg :  the quick brown fox jumps over the lazy dog.

````python
alphabet="abcdefghijklmnopqrstuvxyz"

word="the quick brown fox jumps over the lazy dog."

result=True

for ch in alphabet:

    if ch not in word:

        result=False

        break

print(result)
````

````python
def pangram(word):
    alphabet="abcdefghijklmnopqrstuvxyz"

    result=True

    for ch in alphabet:

        if ch not in word:

            result=False

            break

    print(result)

pangram("the quick brown fox jumps over the lazy dog!")
pangram("Waltz, bad nymph, for quick jigs vex")
````
