## **Multilevel Inheritance**

```python
class GrandParent:

    def properties(self):

        print("GrandParent 50cent land")

class Parent(GrandParent):

    def home(self):

        print("Parent 1000sqft Home")

class Child(Parent):

    def social_media_account(self):

        print("Child Social media accounts")


child_instance = Child()

child_instance.social_media_account()

child_instance.home()

child_instance.properties()
```

## **Ploymorphism**  => **more than one form(Same name but Different behaviour)**



- **method overloading** => same method name but different number of parameter

- **method overriding** => child class redefine methods that are already defined in parent class

```python
class Calculator:

    def add(self,n1,n2):
        print(n1+n2)

    def add(self,n1,n2,n3):

        print(n1+n2+n3)

    def add(self,n1,n2,n3,n4):

        print(n1+n2+n3+n4)

calc_instance = Calculator()

calc_instance.add(1,2,3,4)

```

```python
class Parent:

    def mobile(self):
        print("Vivo X200 fe")

    def bike(self):
        print("rx-100")

class Child(Parent):

    def mobile(self):
        print("iphone 18pro")

    def bike(self):
        print("Passion pro")

child_instance = Child()

child_instance.mobile()

child_instance.bike()
```

**child class have responisibility to call parent class constructor from the child class constructor if the parent class have constructor**  => super()

```python
class Shape:

    name:str

    def __init__(self,name):
        self.name = name

class Square(Shape):

    side:int

    def __init__(self, name,side):

        super().__init__(name)

        self.side = side

    def area(self):

        print("area of ",self.name,self.side**2)

square_instance = Square("Square",12)

square_instance.area()
        
```

```python
class Rectangle(Shape):

    l:int
    w:int

    def __init__(self, name,l,w):
        super().__init__(name)

        self.l = l
        self.w = w

    def area(self):

        print("Area of ",self.name,"is",self.l * self.w)

rec_instance = Rectangle("Rectangle",4,5)

rec_instance.area()
```

```python
class Triangle(Shape):

    b:int
    h:int

    def __init__(self, name,b,h):

        super().__init__(name)

        self.b = b
        self.h = h

    def area(self):
        print("Area of ",self.name,"is",0.5*self.b*self.h)

Tri_instance = Triangle("Triangle",2,3)

Tri_instance.area()


        
```

```python
class Circle(Shape):

    r:int

    def __init__(self, name,r):
        super().__init__(name)

        self.r = r

    def area(self):
        print("Area of",self.name,"is",3.14*self.r**2)

circle_instance = Circle("Circle",2)

circle_instance.area()

```

```python
class Shape:

    name:str

    def __init__(self,name):
        self.name = name

class Parallelogram(Shape):

    b:int
    h:int

    def __init__(self,name,b,h):
        super().__init__(name)

        self.b = b
        self.h = h

    def area(self):
        print("Area of",self.name,"is",self.b*self.h)


para_instance = Parallelogram("Parallelogram",20,13)

para_instance.area()
```

```python
class Trapezium(Shape):

    side1:int
    side2:int
    h:int

    def __init__(self, name,side1,side2,h):
        super().__init__(name)

        self.side1 = side1
        self.side2 = side2

        self.h = h

    def area(self):
        print("Area of",self.name,"is",0.5*(self.side1+self.side2)*self.h)

tra_instance = Trapezium("Trapezium",2,3,4)

tra_instance.area()
```

## **Abstraction** => Hiding Impelementation details showing essential information



```python
from abc import ABC , abstractmethod

class Editor(ABC): #ABC => Abstract Base Class

    @abstractmethod
    def edit(self): pass

    @abstractmethod
    def debug(self): pass

    @abstractmethod
    def execute(self): pass


class PyCharm(Editor):

    def edit(self):
       print("editing")

    def debug(self):
        print("debuggingg....")

    def execute(self):
        print("executing.....")
    

pych_instance = PyCharm()

pych_instance.edit()
```
