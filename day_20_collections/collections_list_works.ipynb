## list methods

- `append(value)` => **Add value to end of the list**

- `insert(pos,value)` => **insert value at sepecified index**

- `pop(index=-1)` => **remove the value at specified index**

````python
daily_calories.append(2100)

daily_calories
````

````python
arr = [10,11,12,13,14,15,16,17]

evens = []

for num in arr:

    if num%2==0:

        evens.append(num)

print(evens)
````

````python
arr = [10,11,12,13,14,15,16,17]

odds = []

for num in arr:

    if num%2!=0:

        odds.append(num)
print(odds)
````

````python
scores = [45,52,51,43,55,57,64,75,23,24]

passed = []

failed = []

for mark in scores:

    if mark>=50:

        passed.append(mark)

    else:

        failed.append(mark)

print("passed:",passed)

print("failed:",failed)




````

````python
#insert(pos,value)

fav_colors = ["red","green","blue"]

fav_colors.insert(0,"olive")

print(fav_colors)
````

````python
#pop()

arr = [10,20,30,40]

arr.pop()  #remove last index =>40
arr.pop(2) #removed at sepecified index =>30

print(arr)  #[10,20]

````
