```python
# read from words.txt and display palindromes

fr = open("words.txt","r")

for line in fr:

    word = line.rstrip("\n")
    if word==word[::-1]:
        print(word)
fr.close()
```

```python
fr = open("words.txt","r")

all_words=[]

for line in fr:
    word = line.rstrip("\n")

    all_words.append(word)

fr.close()

all_words.sort()

first = all_words[0]
last = all_words[-1]

small_word = first if len(first)<len(last) else last

prefix =""

for i in range(len(small_word)):

    if first[i]==last[i]:
        prefix+=first[i]
    else:
        break
print(prefix)
fr.close()
```

```python
# order count

fr = open("orders.txt","r")

all_order=[]

for line in fr:
    item = line.rstrip("\n")

    all_order.append(item)

order_count = {order:all_order.count(order) for order in all_order}

print(order_count)

fr.close()
```

```python
fr = open("wheather.txt","r")

wheather_report=[]

for line in fr:
    line = line.rstrip("\n")

    district,temp=line.split(",")

    dictionary = {"district":district,"temp":temp}

    wheather_report.append(dictionary)

print(wheather_report)
fr.close()
```

```python
fr = open("films.txt","r")

films_report = []

for line in fr:
    line = line.rstrip("\n")

    title,language,rating = line.split(",")

    dictionary = {"title":title,"language":language,"rating":rating}

    films_report.append(dictionary)

print(films_report)

fr.close()
```
